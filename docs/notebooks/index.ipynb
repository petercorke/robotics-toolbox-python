## Kinematics and dynamics
| Notebook | Description |
| :--- | :--- |
| [**Robot kinematics**](./kinematics.ipynb) | Forward and inverse kinematics, a joint-space trajectory and the Jacobian of a Puma 560. |
| [**Elementary transform sequences**](./ets.ipynb) | Build a robot model from the `ET` and `ETS` classes: forward kinematics and Jacobians. |
| [**Working with branched robots**](./branched-robot.ipynb) | A robot with more than one end-effector: kinematics, Jacobians and extracting a branch. |
| [**Robot dynamics**](./dynamics.ipynb) | Inertia, velocity terms and gravity load of a Puma 560, and its forward dynamics. |
| [**Dynamics of a 2-link manipulator**](./two-link-dynamics.ipynb) | Gravity and velocity torques from DH and ETS models, checked against the Spong book. |

## Symbolic and analytic methods
| Notebook | Description |
| :--- | :--- |
| [**Symbolic operations**](./symbolic.ipynb) | Forward kinematics and inverse dynamics with symbolic joint variables, and generating code. |
| [**Analytic inverse kinematics**](./ik-symbolic.ipynb) | Closed-form IK for six-axis arms: all solutions, no initial guess, and generated code in Python, C, C++ and Rust. |
